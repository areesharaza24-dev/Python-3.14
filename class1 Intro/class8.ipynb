- *args = will make sure all the values as extra argument og into it as Tuple

```python
def addition(num1, *num2):
    print(f'*args came as tuple {num2}')
    return num1 + sum(num2)

addition(2,3)
```

```python
def addition(num1, num2, *num3):
    print(f'*args came as Tuple {num3}')
    return num1 + num2 + sum(num3)
addition(2,3,4)
```

```python
def subtraction(num1,num2,*num3):
    print(f'args came as Tuple {num3}')
    return num1 - num2 - sum(num3)
print(subtraction(4,6,8))
```

```python
def testing(*numbers):
    print(*numbers)

testing(2,3,4,5,6)
```

```python
def testing(**kwargs):
    print(fr"*kwargs came as a DICTIONARY {kwargs}")

testing(name='Qasim', height=5.5, weight=40)
```

Q: Create function that print atleat 2 details on customer as paramter and 3 details as **kwargs

```python
def customer_info(name, age, **kwargs):
    print("Customer Details:")
    print("Name:", name)
    print("Age:", age)

    print("Additional Details:")
    for key, value in kwargs.items():
        print(f"{key}:", value)
```

```python
# Example function call
customer_info(
    "Hadi",
    14,
    city="Karachi",
    email="hadi-online@gmail.com",
    phone="0300-9934567"

)
```

```python
for temp in ['Qasim', 'Ahram', 'Hsssan', 'Hadi']:
    print(temp)
```

```python
students = ['Qasim', 'Ahram', 'Hassan', 'Hadi']
is_present = 'No' # check arfa presence in the class

while is_present == 'No': # overall conidition true
    for student in students:
        if student == 'Hassan': # condition true of 3rd loop
            print('Yes Hassan is present')
            is_present = 'Yes'
            break
        else:
            print(f'No, {student} is present')
```

```python
students = ['Qasim', 'Ahram', 'Arfa', 'Hadi']
is_present = 'No'
index = 0

while index < len(students):
    if students[index] == 'Arfa':
        print('Yes Arfa is present')
        is_present = 'Yes'
        break
    else:
        print(f'No, {students[index]} is not Arfa')
    index += 1

if is_present == 'No':
    print('Arfa is not present in the class')
```
