Example:

7- Find the factorial of a number using a for loop.

```python
# Quetion -- 5

sum = 0

for i in range(1,11):
    sum = sum + i
print(sum)
```

```python
# Question -- 6
square = 0
for i in range(1,6):
    square = i * i
    print(square)
```

```python
# Question -- 7
# the factorial of a non-negative integer is the product of all positive integers less than or equal 
# to that number. It is represented by the exclamation mark symbol (!).
num = 5
fact = 1
for i in range(1, num + 1):
    fact = fact * i
print(fact)
```

# Nested loop Practice

- Print the pattern using nested loop


```python
for i in range(5):
    for j in range(5):
        print("*", end= "")
    print()   
```

```python
for i in range(5):
    for j in range(i + 1):
        print("*", end= '')
    print()
```

```python
for i in range(4):
    for j in range(4):
        print("*", end = '')
    print()
```

```python
for i in range(5):
    for j in range(i + 1):
        print(i + 1, end="")
    print()
```

```python
for i in range(5):
    for j in range(i + 1):
        print(j + 1, end="")
    print()
```
