Create a list that contain dictionary

````python
customers = []
````

````python
customer0 = {
    'customer_id':0,
    'first_name':'john',
    'last_name':'organ',
    'address': '301 Arbor Rd',
    'orders' : ['fruits', 'vegetables', 'pulses']
}
````

````python
customer1 = {
    'customer id' : 1,
    'first_name' : 'Qasim',
    'last_name' : 'Hassan',
    'address' : 'xyz rd',
    'orders' : []
}
````

````python
customers.append(customer1)
````

````python
customers.append(customer0)
````

````python
customers
````

How to get information out of a dictionary within a list


````python
customers
````

````python
# print all key value pairs that are in a dictionary with in a list
for customer in customers:
    for key,value in customer.items():
        print(f'{key}--->{value}')
    print('-----')    
````
