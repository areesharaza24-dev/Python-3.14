# Nested Loop

```python
first_names = ['Qasim', 'Muhammad', 'Ali', 'Hussain']
last_names = ['Hassan', 'Raza', 'Khan', 'Bhatti']

for f_name in first_names:
    for l_name in last_names:
        print(f'{f_name} {l_name}')
    print("----")
```

```python
lst1 = [[2,3,4,4],[2,3,4,5]] # 2 length

lst2 = [[6,7,8,9],[3,5,6,2]] # 2 length

result = []
for i in range(len(lst1)):
    # print(lst1[i])
    for j in range(len(lst1[0])):
        # print(lst1[i][j])
        result.append(lst1[i][j] + lst2[i][j])
result
```

# DICTIONARIES

- dictionaries are nothing but key-value pairs

```python
bio_data_list = ['Areesha', 'Raza Siddiqui', 18, 5.2, 'ABC address']

bio_data_dic = {
    'name' : 'Areesha',
    'last_name' : 'Raza Siddiqui',
    'age' : 18,
    'height' : 5.2,
    'address' : 'ABC address'
}

print(bio_data_dic)
```

- Apply for loop in dictionaries


```python
for key in bio_data_dic.keys():
    print(key)
```

```python
for value in bio_data_dic.values():
    print(value)
```

```python
for item in bio_data_dic.items():
    print(item)
```

```python
for key, value in bio_data_dic.items():
    print(f'{key},-- {value}')
```
