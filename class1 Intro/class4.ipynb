 Question: A school has following rules for grading system:

- Below 25 - F
- 25 to 45 - E
- 45 to 50 - D
- 50 to 60 - C
- 60 to 80 - B
- Above 80 - A
- Ask user to enter marks and print the corresponding grade

```python
marks = float(input("Enter marks: "))
if marks > 80:
    print("Your grade is A")
elif marks > 60 and marks <= 80:
    print("Your grade is B")
elif marks > 50 and marks <= 60:
    print("Your grade is C")
elif marks > 45 and marks <= 50:
    print("Your grade is D")
elif marks > 25 and marks <= 45:
    print("Your grade is E")
elif marks < 25:
    print("Your grade is F")
else:
    print("Invalid output")
```

Write a program to check whether a number entered by user is even or odd.

```python
# ============================================ Input =====================================================
number = int(input("Enter a number: "))

# ========================================== CONDITION ===================================================
if number % 2 == 0:
    print("Even Number")

elif number % 2 != 0:
    print("Not Even Number")

else:
    print("Invalid Input")
```

       ============ CONCATENATION ============

```python
fname = "Areesha"
lname = "Raza"
marks = "90"
```

Method-1: Using + Operator

```python
fname + lname
```

```python
fname + " " + lname
```

```python
fname + str(marks) # can only concatenate str (not "int") to str
```

Method -2: using f string

```python
fullname = f"{fname} {lname}"
fullname
```

```python
fullname = f"{fname} {lname}{marks}"
print("your fullname is ",fullname)
```

- NOTE:

- For concatenation using f"" you don't need to TYPE cast.
